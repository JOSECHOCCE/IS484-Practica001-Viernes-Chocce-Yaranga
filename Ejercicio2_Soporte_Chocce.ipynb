## Ejercicio 2: Escalamiento de tickets de soporte técnico

### Ficha PEAS

| Elemento | Descripción |
|---|---|
| **P — Medida de desempeño** | Resolver incidencias críticas rápidamente, reducir el tiempo de espera y asignar correctamente los recursos del equipo de soporte. |
| **E — Entorno** | Mesa de ayuda, usuarios, tickets, técnicos de soporte y sistemas tecnológicos de la organización. |
| **A — Actuadores** | Asignar a nivel 1, escalar a nivel 2 o escalar a nivel 3. |
| **S — Percepciones** | Tiempo de espera en minutos, nivel de urgencia y condición de cliente premium. |
| **Objetivo** | Enviar cada ticket al nivel de soporte adecuado según su prioridad, tiempo de espera y tipo de cliente. |
| **Tipo de agente** | Agente reactivo simple, porque toma decisiones usando reglas y las percepciones actuales del ticket. |

### Justificación de las reglas

Los tickets de urgencia alta se escalan inmediatamente porque pueden afectar servicios importantes o impedir el trabajo del cliente. Los clientes premium reciben atención prioritaria debido al nivel de servicio contratado. Un ticket de urgencia media se escala cuando ha esperado demasiado tiempo o pertenece a un cliente premium. Los demás casos pueden ser atendidos inicialmente por el nivel 1 para usar eficientemente los recursos técnicos.

### Código

```python
def agente_soporte(tiempo_espera_minutos, nivel_urgencia, cliente_premium):
    nivel_urgencia = nivel_urgencia.lower()

    if nivel_urgencia not in ["baja", "media", "alta"]:
        return "ticket invalido", "el nivel de urgencia debe ser baja, media o alta"

    if nivel_urgencia == "alta":
        return "escalar a nivel 3", "urgencia alta reportada; requiere atencion inmediata"

    elif nivel_urgencia == "media" and (
        tiempo_espera_minutos > 60 or cliente_premium
    ):
        return "escalar a nivel 2", "urgencia media con espera elevada o cliente premium"

    elif nivel_urgencia == "baja" and cliente_premium and tiempo_espera_minutos > 120:
        return "escalar a nivel 2", "cliente premium con baja urgencia y espera mayor a 120 minutos"

    else:
        return "asignar a nivel 1", "ticket que puede ser atendido inicialmente por soporte de primer nivel"
```

### Simulación y pruebas

```python
casos_soporte = [
    (10, "alta", False),     # Urgencia alta: nivel 3
    (30, "media", False),    # Urgencia media con espera normal: nivel 1
    (80, "media", False),    # Urgencia media y espera mayor a 60 min: nivel 2
    (20, "media", True),     # Urgencia media con cliente premium: nivel 2
    (130, "baja", True),     # Caso especial: baja + premium + espera alta: nivel 2
    (150, "baja", False),    # Baja urgencia sin cliente premium: nivel 1
]

for espera, urgencia, premium in casos_soporte:
    accion, motivo = agente_soporte(espera, urgencia, premium)

    print(f"Tiempo de espera: {espera} minutos")
    print(f"Nivel de urgencia: {urgencia}")
    print(f"Cliente premium: {premium}")
    print(f"Decisión: {accion}")
    print(f"Motivo: {motivo}")
    print("-" * 70)
```
